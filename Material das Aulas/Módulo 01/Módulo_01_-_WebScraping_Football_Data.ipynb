```python
# Importando as Bibliotecas

import pandas as pd
import numpy as np

import warnings
warnings.filterwarnings('ignore')

def drop_reset_index(df):
    df = df.dropna()
    df = df.reset_index(drop=True)
    df.index += 1
    return df
```

```python
url = 'https://www.football-data.co.uk/mmz4281/2324/E0.csv'

df = pd.read_csv(url)
```

```python
df.columns.to_list()
```

```python
colunas = ['Date','HomeTeam','AwayTeam','FTHG','FTAG','FTR','B365H','B365D','B365A','B365>2.5','B365<2.5']
colunas_renomeadas = ['Date','Home','Away','Goals_H','Goals_A','Result','Odd_H','Odd_D','Odd_A','Odd_Over25','Odd_Under25']

df = df[colunas]
df.columns = colunas_renomeadas

df = drop_reset_index(df)

display(df)
```

```python
df.to_excel("./Football_Data/Base_Football_Data.xlsx",index=False)
```

```python
# Pegando de Várias Ligas

df1 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2122/F1.csv')
df2 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2223/F1.csv')
df3 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2324/F1.csv')
df4 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2122/I1.csv')
df5 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2223/I1.csv')
df6 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2324/I1.csv')
df7 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2122/D1.csv')
df8 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2223/D1.csv')
df9 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2324/D1.csv')
df10 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2122/SP1.csv')
df11 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2223/SP1.csv')
df12 = pd.read_csv('https://www.football-data.co.uk/mmz4281/2324/SP1.csv')

df = pd.concat([df1,df2,df3,df4,df5,df6,df7,df8,df9,df10,df11,df12])
df = df[['Date','HomeTeam','AwayTeam','FTHG','FTAG','B365H','B365D','B365A']]
df.columns = ['Date','Home','Away','Goals_H','Goals_A','Odd_H','Odd_D','Odd_A']

df = drop_reset_index(df)
df.to_excel("./Football_Data/Base_Football_Data.xlsx",index=False)
display(df)
```
