### Jogos do Dia

```python
def fetch_matches(date, key, pages=3):
    dfs = []
    for page in range(1, pages + 1):
        url = f"https://api.football-data-api.com/todays-matches?key={key}&date={date}&page={page}"
        response = requests.get(url).json().get('data', [])
        if response:  
            temp_df = pd.DataFrame.from_dict(response)
            if 'competition_id' in temp_df.columns:
                dfs.append(temp_df)
    return pd.concat(dfs) if dfs else pd.DataFrame()

def process_matches(dates, key):
    all_matches = pd.concat([fetch_matches(date.strftime("%Y-%m-%d"), key) for date in dates], ignore_index=True)
    if not all_matches.empty and 'competition_id' in all_matches.columns:
        filtered_columns = [
            'competition_id', 'date_unix', 'game_week', 'home_name', 'away_name', 
            'odds_1st_half_result_1', 'odds_1st_half_result_x', 'odds_1st_half_result_2', 
            'odds_ft_1', 'odds_ft_x', 'odds_ft_2', 'odds_1st_half_over05', 'odds_1st_half_under05', 
            'odds_ft_over05', 'odds_ft_under05', 'odds_ft_over15', 'odds_ft_under15', 
            'odds_ft_over25', 'odds_ft_under25', 'odds_btts_yes', 'odds_btts_no', 
            'odds_doublechance_1x', 'odds_doublechance_12', 'odds_doublechance_x2', 
            'home_ppg', 'away_ppg', 'team_a_xg_prematch', 'team_b_xg_prematch', 'total_xg_prematch', 
            'odds_corners_1', 'odds_corners_x', 'odds_corners_2', 'odds_corners_over_75', 
            'odds_corners_under_75', 'odds_corners_over_85', 'odds_corners_under_85', 
            'odds_corners_over_95', 'odds_corners_under_95', 'odds_corners_over_105', 
            'odds_corners_under_105', 'odds_corners_over_115', 'odds_corners_under_115'
        ]
        all_matches = all_matches.loc[:, all_matches.columns.intersection(filtered_columns)]
        if 'date_unix' in all_matches.columns:
            all_matches['Data'] = pd.to_datetime(all_matches['date_unix'], unit='s') - pd.DateOffset(hours=3)
        return all_matches
    else:
        return pd.DataFrame()  

dates = [date.today() + timedelta(days=i) for i in range(3)]  
df0 = process_matches(dates, key)
```

```python
df = df0[['competition_id','Data','game_week','home_name','away_name','odds_ft_1','odds_ft_x','odds_ft_2','odds_ft_over05','odds_ft_under05','odds_ft_over15','odds_ft_under15','odds_ft_over25','odds_ft_under25','odds_btts_yes','odds_btts_no','odds_doublechance_1x','odds_doublechance_12','odds_doublechance_x2']]
df.columns = ['League','Data','Rodada','Home','Away','Odd_H','Odd_D','Odd_A','Odd_Over05','Odd_Under05','Odd_Over15','Odd_Under15','Odd_Over25','Odd_Under25','Odd_BTTS_Yes','Odd_BTTS_No','Odd_DC_1X','Odd_DC_12','Odd_DC_X2']
df = drop_reset_index(df)
rename_leagues(df)

df = df.sort_values(['Data'])
df = df.drop_duplicates()
df = drop_reset_index(df)

df['Date'] = df['Data'].dt.date
df['Time'] = df['Data'].dt.time
df = df[['League','Date','Time','Rodada','Home','Away','Odd_H','Odd_D','Odd_A','Odd_Over05','Odd_Under05','Odd_Over15','Odd_Under15','Odd_Over25','Odd_Under25','Odd_BTTS_Yes','Odd_BTTS_No','Odd_DC_1X','Odd_DC_12','Odd_DC_X2']]
df
```

```python
def convert_date_to_datetime(date_obj):
    if isinstance(date_obj, date) and not isinstance(date_obj, datetime):
        return datetime.combine(date_obj, time.min)
    return date_obj

def convert_time_to_string(time_obj):
    if isinstance(time_obj, time):
        return time_obj.strftime('%H:%M:%S')
    return time_obj

def fetch_and_process_matches(league, season_id, year, key, db_collection):
    url = f"https://api.football-data-api.com/league-matches?key={key}&season_id={season_id}"
    response = requests.get(url).json()['data']
    df = pd.DataFrame.from_dict(response)
    df['League'] = league
    # Selecione e renomeie colunas conforme necessário
    df = df[['id','League','season','status','date_unix','game_week','home_name','away_name',
                 'homeGoalCount','awayGoalCount','totalGoalCount',
                 'homeGoals','awayGoals',
                 'odds_ft_1','odds_ft_x','odds_ft_2',
                 'odds_ft_over05','odds_ft_under05','odds_ft_over15','odds_ft_under15','odds_ft_over25','odds_ft_under25',
                 'odds_btts_yes','odds_btts_no','odds_doublechance_1x','odds_doublechance_12','odds_doublechance_x2']]
    df.columns = ['Id_Jogo','League','Season','Status','Data','Rodada','Home','Away',
                      'Goals_H','Goals_A','TotalGoals',
                      'Goals_H_Minutes','Goals_A_Minutes',
                      'Odd_H','Odd_D','Odd_A',
                      'Odd_Over05','Odd_Under05','Odd_Over15','Odd_Under15','Odd_Over25','Odd_Under25',
                      'Odd_BTTS_Yes','Odd_BTTS_No','Odd_DC_1X','Odd_DC_12','Odd_DC_X2']

    df['Data'] = pd.to_datetime(df['Data'], unit='s') - pd.DateOffset(hours=3)
    df = df.sort_values(['Data'])
    df = df[df.Status == 'complete']
    df['Date'] = df['Data'].dt.date
    df['Time'] = df['Data'].dt.time
    df = df[['League','Season','Date','Time','Rodada','Home','Away',
             'Goals_H','Goals_A','TotalGoals',
             'Goals_H_Minutes','Goals_A_Minutes',
             'Odd_H','Odd_D','Odd_A',
             'Odd_Over05','Odd_Under05','Odd_Over15','Odd_Under15','Odd_Over25','Odd_Under25',
             'Odd_BTTS_Yes','Odd_BTTS_No','Odd_DC_1X','Odd_DC_12','Odd_DC_X2']]

    data_dict = df.to_dict("records")

    for record in data_dict:
        for key, value in record.items():
            value_converted_date = convert_date_to_datetime(value) 
            record[key] = convert_time_to_string(value_converted_date) 

    db_collection.insert_many(data_dict)
    
    print(f"{league} - {year} - Salvo no MongoDB com Sucesso!")

client = MongoClient("mongodb://localhost:27017/")  
db = client['Bases_de_Dados']  
collection = db['FootyStats']  

for i, j, k in zip(new2.League, new2.ID, new2.Season):
    try:
        fetch_and_process_matches(str(i), str(j), str(k), key, collection)
    except:
        pass
```
