```python
import pandas as pd
import numpy as np
import ast

import warnings
warnings.filterwarnings('ignore')

def drop_reset_index(df):
    df = df.dropna()
    df = df.reset_index(drop=True)
    df.index += 1
    return df

base_orig = pd.read_csv("https://github.com/futpythontrader/YouTube/raw/main/Bases_de_Dados/FlashScore/Base_de_Dados_FlashScore_v2.csv")
base_orig["Date"] = pd.to_datetime(base_orig["Date"])
base_orig = base_orig.sort_values('Date')

def convert_string_to_list(string):
    if pd.isnull(string) or string == "[]":
        return []
    else:
        try:
            result = ast.literal_eval(string)
            if isinstance(result, int):
                return [result]
            return result
        except ValueError:
            return []

def goal_order_and_minutes(home_goals, away_goals):
    home_goals = home_goals if isinstance(home_goals, list) else []
    away_goals = away_goals if isinstance(away_goals, list) else []

    combined_goals = sorted([(goal, 'H') for goal in home_goals] + [(goal, 'A') for goal in away_goals], key=lambda x: x[0])

    first_goal, second_goal = None, None
    first_goal_minute, second_goal_minute = None, None

    if len(combined_goals) > 0:
        first_goal_minute, first_goal = combined_goals[0]
    if len(combined_goals) > 1:
        second_goal_minute, second_goal = combined_goals[1]

    return first_goal, second_goal, first_goal_minute, second_goal_minute

base_orig['Goals_Minutes_Home'] = base_orig['Goals_Minutes_Home'].apply(convert_string_to_list)
base_orig['Goals_Minutes_Away'] = base_orig['Goals_Minutes_Away'].apply(convert_string_to_list)

base_orig[['First_Goal', 'Second_Goal', 'First_Goal_Minute', 'Second_Goal_Minute']] = base_orig.apply(
    lambda row: goal_order_and_minutes(row['Goals_Minutes_Home'], row['Goals_Minutes_Away']),
    axis=1, result_type='expand'
)
```
